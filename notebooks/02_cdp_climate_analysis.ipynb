At Walmart, ESG and shared value initiatives are led by the CEO and executive leadership team. The Chief Sustainability Officer (CSO) helps define the ESG agenda and provides dedicated management oversight of global ESG goals and initiatives.

The frequency with which senior management reports sustainability issues to the Board provides an indication of how formally ESG oversight is embedded within governance processes.

Walmart's Chief Sustainability Officer updates and receives feedback from the executive leadership team and the Board's Nominating and Governance Committee on ESG strategies and priorities at least annually.

The disclosure therefore provides evidence of a recurring formal reporting mechanism between sustainability management and Board level oversight.

### Physical and transition climate risks

Walmart distinguishes between two broad categories of climate related risk.

**Physical risks** arise from the physical effects of climate change, including acute events such as extreme weather and longer term changes in climate patterns. These risks may disrupt Walmart's operations and supply chains.

**Transition risks** arise from the economic transition toward a lower carbon economy and include changes in regulation, policy, technology and market conditions.

Walmart identifies both categories as potentially relevant to its business and incorporates them into its climate risk assessment.

### Climate scenario analysis

Walmart uses climate scenario analysis to examine how its business could be affected under different potential climate futures.

For physical climate risks, Walmart applied the **IPCC AR6 SSP5-8.5** scenario, representing a high emissions pathway.

For transition risks, Walmart applied the **NGFS Net Zero 2050** scenario, representing a rapid transition toward a net-zero economy.

Using contrasting scenarios allows the assessment to test exposure to both substantial physical climate change and substantial economic and regulatory transition.

### Time horizons

Walmart evaluates climate related risks across multiple time horizons:

- **Short term:** 1 year
- **Medium term:** 5 years
- **Long term:** 15 years

The short and medium term horizons are aligned with Walmart's annual and longer range business planning cycles.

Using multiple time horizons allows climate risks to be considered alongside both immediate operational decisions and longer term strategic planning.

```python
walmart_risk_horizons = {
    "short_term": 1,
    "medium_term": 5,
    "long_term": 15
}

walmart_risk_horizons
```

### Integration into enterprise risk management

Walmart does not assess climate risk solely as a standalone sustainability issue.

Climate related risks are assessed at least annually through the company's broader **Enterprise Risk Management (ERM)** process. Outputs from this process are discussed with management level governance bodies, while the Board's Audit Committee oversees the company's overall risk identification, monitoring and mitigation processes.

This indicates that climate related risk assessment is connected to Walmart's wider corporate risk management architecture rather than being confined to ESG reporting.

#### Physical climate risk

Walmart identifies physical climate change as a potential source of disruption to both its own operations and its wider value chain.

For a retailer, the significance of physical climate risk extends beyond direct damage to stores or distribution infrastructure. Climate related disruption can propagate through the supply chain and ultimately affect the availability and cost of products sold to customers.

A simplified risk pathway can therefore be represented as:

**Climate hazard → operational or supplier disruption → reduced product availability / increased costs → potential business impact**

This is particularly relevant for agricultural and commodity dependent supply chains, where production may be sensitive to extreme weather and changing climatic conditions.

### Climate adaptation and resilience

Identifying climate risks does not by itself demonstrate resilience. The next question is therefore whether Walmart describes actions intended to reduce exposure or improve its ability to respond to climate related disruption.

Walmart describes several approaches relevant to supply chain resilience, including:

- diversification of suppliers;
- local and regional sourcing;
- use of technology and data;
- sustainable commodity sourcing;
- inventory management; and
- broader measures intended to maintain continuity of product supply.

These measures do not necessarily eliminate climate risk. Instead, they may reduce Walmart's dependence on individual suppliers, production regions or supply pathways and improve its ability to respond when disruption occurs.

#### Supplier diversification as a resilience mechanism

Supplier diversification can reduce concentration risk within the supply chain.

If a company relies heavily on a single supplier or production region, a climate related disruption in that location may have a disproportionate impact on product availability.

Diversification can spread this exposure across suppliers and geographies.

However, diversification should not automatically be interpreted as eliminating climate risk. Multiple suppliers may remain exposed to correlated hazards, particularly where production is geographically concentrated or dependent on similar climatic conditions.

### Financial materiality of climate-related risk

Identifying a climate hazard does not necessarily demonstrate that a company understands its potential financial significance.

A more developed climate risk assessment considers the pathways through which climate related events could affect financial performance, such as:

- reduced revenue;
- increased operating costs;
- asset damage;
- supply shortages;
- business interruption; and
- changes in energy or input costs.

Walmart's 2024 climate risk assessment identifies potential financial impacts associated with individual physical climate risks.
